```python
import sys

# Clean up any broken references to torch to fix the circular import state
for key in list(sys.modules.keys()):
    if 'torch' in key or 'modelscope' in key or 'paddlex' in key or 'paddleocr' in key:
        sys.modules.pop(key, None)

# Initialize torch cleanly first to establish its namespace
import torch
import paddle
import paddleocr

print("PaddlePaddle:", paddle.__version__)
print("PaddleOCR:", paddleocr.__version__)
print("GPU available:", paddle.is_compiled_with_cuda())
print("Device:", paddle.device.get_device())
```

```python
from google.colab import files

uploaded = files.upload()
input_file = next(iter(uploaded))

print("Uploaded:", input_file)
```

```python
from paddleocr import PaddleOCRVL
from pathlib import Path

output_dir = Path("/content/ocr_output")
output_dir.mkdir(exist_ok=True)

# Fallback to CPU to avoid GPU Out-Of-Memory (OOM) issues on T4 with the 0.9B VLM model
pipeline = PaddleOCRVL(device="cpu")

results = pipeline.predict(input=input_file)

for res in results:
    res.print()
    res.save_to_json(save_path=str(output_dir))
    res.save_to_markdown(save_path=str(output_dir))

print("OCR completed!")
```
